```python
import torch.nn as nn
```

```python
class LinearModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear= nn.Linear(1,1)

    def forward(self,x):
        return self.linear(x)

model = LinearModel()
loss_fn= nn.MSELoss()
optimizer=torch.optim.SGD(model.parameters(),lr=0.1)

for epoch in range(100):
    y_pred = model(x)
    loss=loss_fn(y_pred,y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()


    if epoch%20==0:
        print(epoch,loss.item())
```

```python
seq_len, d_model, num_heads, num_kv_heads = 5, 8, 4, 2
d_k = d_model // num_heads
group_size = num_heads // num_kv_heads   # how many q heads share each kv head

x = torch.randn(seq_len, d_model)

W_q = nn.Linear(d_model, num_heads * d_k)        # = nn.Linear(8, 8)
W_k = nn.Linear(d_model, num_kv_heads * d_k)     # = nn.Linear(8, 4)  <- smaller
W_v = nn.Linear(d_model, num_kv_heads * d_k)     # = nn.Linear(8, 4)  <- smaller
W_o = nn.Linear(d_model, d_model)

Q = W_q(x)   # (5, 8)
K = W_k(x)   # (5, 4)
V = W_v(x)   # (5, 4)

# TODO 1: reshape Q into (num_heads, seq_len, d_k) -- same as your MHA code
Q = Q.view(seq_len, num_heads, d_k).transpose(0,1)


# TODO 2: reshape K, V into (num_kv_heads, seq_len, d_k)
K = K.view(seq_len, num_kv_heads, d_k).transpose(0,1)
V = V.view(seq_len, num_kv_heads, d_k).transpose(0,1)


# TODO 3: repeat K and V along the head dimension so they go from
# num_kv_heads -> num_heads (each kv head repeated `group_size` times)
# hint: look up torch.repeat_interleave, repeat along dim=0
K=torch.repeat_interleave(K,group_size,dim=0)
V=torch.repeat_interleave(V,group_size,dim=0)
# after this, K and V should both be shape (num_heads, seq_len, d_k), matching Q

mask = torch.triu(torch.ones(seq_len, seq_len), diagonal=1).bool()
scores = Q @ K.transpose(-2,-1) / (d_k ** 0.5)
scores = scores.masked_fill(mask, float('-inf'))
weights = torch.softmax(scores, dim=-1)
attn_out = weights @ V   # (num_heads, seq_len, d_k)

# TODO 4: same as MHA -- transpose back, reshape/flatten to (seq_len, d_model), then W_o
output = attn_out.transpose(0,1).reshape(seq_len,d_model)
output=W_o(output)
print(output.shape)  # should be (5, 8)
```

```python
class GQA(nn.Module):
    def __init__(self, d_model, num_heads, num_kv_heads):
        super().__init__()
        self.d_model = d_model
        self.num_heads = num_heads
        self.num_kv_heads = num_kv_heads
        self.d_k = d_model // num_heads
        self.group_size = num_heads // num_kv_heads

        # TODO 1: declare W_q, W_k, W_v, W_o as nn.Linear layers here
        # (same in_features/out_features math as before)
        self.W_q = nn.Linear(d_model,d_k*num_heads)
        self.W_k = nn.Linear(d_model,d_k*num_kv_heads)
        self.W_v = nn.Linear(d_model,d_k*num_kv_heads)
        self.W_o = nn.Linear(d_model,d_model)

    def forward(self, x):
        seq_len = x.shape[0]

        Q = self.W_q(x)
        K = self.W_k(x)
        V = self.W_v(x)

        # TODO 2: reshape+transpose Q, K, V (use self.num_heads, self.num_kv_heads, self.d_k)
        Q = Q.view(seq_len, self.num_heads, self.d_k).transpose(0,1)
        K = K.view(seq_len, self.num_kv_heads, self.d_k).transpose(0,1)
        V = V.view(seq_len, self.num_kv_heads, self.d_k).transpose(0,1)


        # TODO 3: repeat_interleave K, V using self.group_size
        K = torch.repeat_interleave(K, self.group_size, dim=0)
        V = torch.repeat_interleave(V, self.group_size, dim=0)

        mask = torch.triu(torch.ones(seq_len, seq_len), diagonal=1).bool()
        scores = Q @ K.transpose(-2, -1) / (self.d_k ** 0.5)
        scores = scores.masked_fill(mask, float('-inf'))
        weights = torch.softmax(scores, dim=-1)
        attn_out = weights @ V

        # TODO 4: transpose back, reshape, W_o
        output = attn_out.transpose(0,1).reshape(seq_len, self.d_model)
        output = self.W_o(output)
        return output
```

```python
model=GQA(d_model=8,num_heads=4,num_kv_heads=2)
x=torch.randn(5,8)
out=model(x)
print(out.shape)
```
